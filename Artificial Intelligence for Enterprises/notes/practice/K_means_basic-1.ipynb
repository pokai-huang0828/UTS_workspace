**Silhouette**:
using this approch, you can find the natural number of clusters. in a nutshel, silhouette of each point shows how closely it is matched to data within its cluster. Silhouette mean is between (-1,1). the more closets to one shows the chosen value of k is a good one. 
So in a simple term, you can compute silhouette metric for various values of k and the highest silhouette locate the optimum number of clusters. 

lest first see how you can compute the silhouette values. 

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples, silhouette_score
import numpy as np
X = np.array([[1, 2], [1, 4], [1, 0], [10, 2], [10, 4], [10, 0]])
kmeans = KMeans(n_clusters=2, random_state=0).fit(X)
silhouette_avg = silhouette_score(X, kmeans.labels_)
silhouette_avg
```

now lets see whether silhouette metric is usefull using a synthetic data: (1,2); (1,4); (1,0); (10,2); (10,4); (10,0); (20, 1); (20, 2); (20,4). As you can see the optimum value of k is 3, so the silhouette value of k=3 should be higehr than the one for k=2. 

```python
# silhouette metric for k=2
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples, silhouette_score
import numpy as np
X = np.array([[1, 2], [1, 4], [1, 0], [10, 2], [10, 4], [10, 0],[20, 1], [20, 2], [20, 4] ])
kmeans = KMeans(n_clusters=2, random_state=0).fit(X)
silhouette_avg = silhouette_score(X, kmeans.labels_)
silhouette_avg
```

```python
# silhouette metric for k=3
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples, silhouette_score
import numpy as np
X = np.array([[1, 2], [1, 4], [1, 0], [10, 2], [10, 4], [10, 0],[20, 1], [20, 2], [20, 4] ])
kmeans = KMeans(n_clusters=3, random_state=0).fit(X)
silhouette_avg = silhouette_score(X, kmeans.labels_)
silhouette_avg
```

*   The silhouette value for k=3 is 0.73
*   the silhouette value for k=2 is 0.62
