```python
import pandas as pd
import numpy as np 
from sklearn.metrics import silhouette_samples, silhouette_score
from sklearn.cluster import KMeans
df= pd.read_csv('BigBlue.csv') 
df.head()
```

Droping the varible, xxx,  which is not helpfull for the clustering! 

```python
df_feature= df.drop(['xxx'], axis=1)
df_feature.head()
```

**You may first explore the basic feature of the dataset!**

```python
df.info()
```

You may use silhouette_score for finding the optimum number of clusters! 

```python
silh=[]
silh = [0 for i in range(10)] 

y= range(2,10)
for n in y:
 kmeans = KMeans(n_clusters=n).fit(df_feature)
 silh[n] = silhouette_score(df_feature, kmeans.labels_)
print(silh)
np.argmax(silh)
```

you can use below code to see the cluster centers, I did for the case when you decide the value of 4 as number of clusters

```python
kmeans = KMeans(n_clusters=4).fit(df_feature)
kmeans.cluster_centers_
```

```python
np.around(kmeans.cluster_centers_, decimals=6)
```
